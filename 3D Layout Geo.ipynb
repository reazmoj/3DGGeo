```python
import os
import shutil
import random

# Paths
dataset_root = 'D:/Hashemi/3D Geo v3/dataset/new_dataset'   # <-- change this to your dataset path
train_output = 'D:/Hashemi/3D Geo v3/dataset/train_test'          # <-- change this to your desired train output path
num_folders = 400

# Get all subfolders
all_subfolders = [f for f in os.listdir(dataset_root) if os.path.isdir(os.path.join(dataset_root, f))]

# Randomly select 400 subfolders
selected_folders = random.sample(all_subfolders, min(num_folders, len(all_subfolders)))

# Create train output folder if not exist
os.makedirs(train_output, exist_ok=True)

# Copy selected folders
for folder in selected_folders:
    src = os.path.join(dataset_root, folder)
    dst = os.path.join(train_output, folder)
    shutil.copytree(src, dst)

print(f"✅ Done. Copied {len(selected_folders)} subfolders to '{train_output}'.")
```

```python
import os
import shutil
import random

dataset_root = 'D:/Hashemi/3D Geo v3/dataset/train_test/train'
val_output = 'D:/Hashemi/3D Geo v3/dataset/train_test/test'
val_ratio = 0.2

query_dir = os.path.join(val_output, 'query')
gallery_dir = os.path.join(val_output, 'gallery')
os.makedirs(query_dir, exist_ok=True)
os.makedirs(gallery_dir, exist_ok=True)

group_ids = [d for d in os.listdir(dataset_root) if os.path.isdir(os.path.join(dataset_root, d))]
random.shuffle(group_ids)
val_count = int(len(group_ids) * val_ratio)
val_groups = group_ids[:val_count]

for group in val_groups:
    group_path = os.path.join(dataset_root, group)
    images = [f for f in os.listdir(group_path) if f.endswith(('.jpg', '.png'))]
    if len(images) < 2:
        continue  # skip groups with fewer than 2 images
    random.shuffle(images)

    query_img = images[0]
    # move query
    os.makedirs(os.path.join(query_dir,group), exist_ok=True)
    os.makedirs(os.path.join(query_dir,group + "/depth"), exist_ok=True)
    shutil.move(os.path.join(group_path, query_img), os.path.join(query_dir , f"{group}/{query_img}"))
    label_path = os.path.splitext(query_img)[0] + ".txt"
    shutil.move(os.path.join(group_path, label_path), os.path.join(query_dir, f"{group}/{label_path}"))
    depth_path = os.path.splitext(query_img)[0] + ".npy"
    shutil.move(os.path.join(group_path + "/depth", depth_path), os.path.join(query_dir, f"{group}/depth/{depth_path}"))

    # move gallery
    for gallery_img in images[1:]:
        os.makedirs(os.path.join(gallery_dir,group), exist_ok=True)
        os.makedirs(os.path.join(gallery_dir,group + "/depth"), exist_ok=True)
        shutil.move(os.path.join(group_path, gallery_img), os.path.join(gallery_dir, f"{group}/{gallery_img}"))
        label_path = os.path.splitext(gallery_img)[0] + ".txt"
        shutil.move(os.path.join(group_path, label_path), os.path.join(gallery_dir, f"{group}/{label_path}"))
        depth_path = os.path.splitext(gallery_img)[0] + ".npy"
        shutil.move(os.path.join(group_path + "/depth", depth_path), os.path.join(gallery_dir, f"{group}/depth/{depth_path}"))

    shutil.rmtree(group_path)

print(f"✅ Validation split done. Query and gallery images prepared from {val_count} groups.")
```
